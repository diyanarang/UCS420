```python
import pandas as pd
```

```python
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]
```

```python
last2=list(map(int,last2))
print(last2)
```

```python
categories = ["billing", "account", "general"]

personalized_entries = []
```

```python
for d in last2:
  category = categories[d % 3]
  if category == "general":
        question = "where can i find college information"
        answer = "college information is available on the official website."
        keywords = "college information website details"
  elif category == "billing":
        question = "how can i check my fee payment status"
        answer = "through the student portal."
        keywords = "fee payment status"
  elif category == "account":
        question = "how do i update my registered mobile number"
        answer = "through your account settings."
        keywords = "mobile number update account"
  personalized_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })
```

```python
df = pd.DataFrame(fixed_entries + personalized_entries)
print(df)
```

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():

        text = (
            row["question"] + " " +
            row["keywords"] + " " +
            row["category"]
        ).lower()

        text_words = set(text.split())

        matches = query_words.intersection(text_words)

        if matches:
            confidence = len(matches) / len(query_words)

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": round(confidence, 2)
            })

    results.sort(key=lambda x: x["confidence"], reverse=True)

    return results
```

```python
query = input("Enter your query: ")

results = score_query(query, df)

for result in results:
    print(result)
```

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():

        text = (
            row["question"] + " " +
            row["keywords"] + " " +
            row["category"]
        ).lower()

        text_words = set(text.split())

        matches = query_words.intersection(text_words)

        if matches:
            confidence = len(matches) / len(query_words)

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": round(confidence, 2)
            })

    results.sort(key=lambda x: x["confidence"], reverse=True)

    if not results:
        print("No matching entries found.")
        return

    highest_score = results[0]["confidence"]

    best_matches = [
        result for result in results
        if result["confidence"] == highest_score
    ]

    print("Highest confidence:", highest_score)

    if len(best_matches) > 1:
        print("\nTie! Multiple entries have the same highest score:")

    for result in best_matches:
        print(result)
```

```python
score_query("fee", df)
score_query("reset password", df)
```
