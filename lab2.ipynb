```python
def add(a,b):
  c=a+b
  return c

print(add(2,3))
print(add(6,7))
```

```python
def is_prime(n):
  for i in range (2,int(n**(0.5)+1)):
    if n%i==0:
      return False
  return True

print(is_prime(2))
print(is_prime(87))
```

```python
n=int(input("enter number: "))
def add_odd(n):
  s=0
  for i in range(1,n+1):
    if i%2!=0:
      s=s+i
  return s
add_odd(n)
```
