```python
import random
roll_number = int(input("Enter your roll number: "))
```

```python
random.seed(roll_number)

numbers = [random.randint(100, 900) for _ in range(100)]
print("Random numbers:", numbers)
```

```python
odd_numbers = [num for num in numbers if num % 2 != 0]
print("Odd numbers:", odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
```

```python
even_numbers = [num for num in numbers if num % 2 == 0]
print("Even numbers:", even_numbers)
print("Count of even numbers:", len(even_numbers))
```

```python
def is_prime(num):
    if num < 2:
        return False

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False

    return True

prime_numbers = [num for num in numbers if is_prime(num)]

print("Prime numbers:", prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
```

```python
most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print("Most frequent number:", most_frequent)
print("Number of times it occurs:", frequency)
```

```python
roll_number = input("Enter your roll number: ")
digits = [int(d) for d in roll_number[:8]]
```

```python
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print(A)
print(B)
```

```python
print("Union",A.union(B))
```

```python
print("Intersection", A.intersection(B))
```

```python
print("A-B:", A.difference(B))
print("B-A:", B.difference(A))
# difference() gives elements in one set but not the other,while symmetric_difference() gives elements that are in either set but not both.
```

```python
print("Symmetric difference:", A.symmetric_difference(B))
```

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

```python
X = int(input("Enter a value to remove from A:"))
A.discard(X)
print("A after discard:", A)

# discard() is safer because it does not give an error if the value is not present.
```
