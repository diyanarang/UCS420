```python
df = pd.read_csv("employees.csv")
print(df)
```

```python
print("Shape:", df.shape)
```

```python
print("\nDataFrame Information:")
print(df.info())
```

```python
print("\nDescriptive Statistics:")
print(df.describe())
```

```python
print("\nFirst 5 rows:")
print(df.head(5))

print("\nLast 3 rows:")
print(df.tail(3))
```

```python
# i. Average salary
print("\nAverage Salary:", df["Salary"].mean())

# ii. Total bonus
print("Total Bonus:", df["Bonus"].sum())

# iii. Youngest employee
print("Youngest Age:", df["Age"].min())

# iv. Highest performance rating
print("Highest Rating:", df["Rating"].max())
```

```python
df = df.sort_values(by="Salary", ascending=False)

print("\nSorted by Salary:")
print(df)
```

```python
def performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance"] = df["Rating"].apply(performance)

print("\nPerformance Category:")
print(df[["Name", "Rating", "Performance"]])
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df = df.rename(columns={"Employee_ID": "ID"})

print("\nAfter Renaming:")
print(df)
```

```python
#Find employees with more than 5 years experience
print("\nEmployees with more than 5 years experience:")
print(df[df["Years_of_Experience"] > 5])

# Employees belonging to IT department
print("\nEmployees in IT department:")
print(df[df["Department"] == "IT"])
```

```python
df["Tax"] = df["Salary"] * 0.10

print("\nAfter Adding Tax:")
print(df)
```

```python
df.to_csv("modified_employees.csv", index=False)

print("\nModified dataset saved as modified_employees.csv")
```
