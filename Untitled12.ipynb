```python
steps=np.array([[5000,6200,7100],[8000,7500,9000],[4500,5100,4800],[9000,8500,9500]])

print("Total steps:",np.sum(steps))
print("Mean steps:",np.mean(steps))
print("Maximum:",np.max(steps))
print("Minimum:",np.min(steps))
print("Total steps for each day:",np.sum(steps,axis=0))
print("Total steps for each user:",np.sum(steps,axis=1))
print("Position of maximum:",np.unravel_index(np.argmax(steps),steps.shape))
```

```python
y=np.array([40,65,30,85])
X=np.array([[6,70,3],[5,50,6],[8,80,2],[4,30,8]])
#a
print("shape:",X.shape)
print("dimensions:",X.ndim)
#b
print("transpose:")
print(X.T)
#c
XtX=X.T@X
print("X.T @ X:")
print(XtX)
#d
inverse=np.linalg.inv(XtX)
print("inverse:")
print(inverse)
#e
beta=inverse@X.T@y
print("coefficients:",beta)
```

We get:

Sleep     = -0.159
Activity  =  0.141
Stress    = 10.063

These coefficients represent how each feature contributes to the predicted assistance score while considering the other features in the model.

```python
#g
new_user=np.array([5,40,7])
prediction=new_user@beta
print("Predicted assistance score:",prediction)
```
